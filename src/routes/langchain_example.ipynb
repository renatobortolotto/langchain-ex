```python
import os 
from langchain_google_vertexai import VertexAI
from langchain_core.prompts import PromptTemplate
from langcahin_core.output_parsers import StrOutputParser
import vertexai

from genai_framework.decorators import post_route

@post_route('lanchain_traducao_linguas')
def multimodal_translation(user_input: str):
    vertexai.init(project=os.getenv('GOOGLE_CLOUD_PROJECT'), location=os.getenv('GOOGLE_CLOUD_REGION'))
    llm = VertexAI(
                   model_name='gemini-1.5-pro',
                    temperature=0.7)
    
    prompt = PromptTemplate(
        input_variables=['raw_text'],
        template='Translate the following text into English: {user_input}'
    )
    
    chain = prompt | llm | StrOutputParser()

    return chain.invoke(("raw_text", user_input))
```
